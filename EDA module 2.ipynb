```python
import pandas as pd

i = pd.MultiIndex.from_tuples([
    ('Karnataka', 2010), ('Karnataka', 2020),
    ('Maharashtra', 2010), ('Maharashtra', 2020),
    ('Tamil Nadu', 2010), ('Tamil Nadu', 2020)
])

s = pd.Series([60, 68, 110, 125, 72, 78], index=i)
print(s)
```

```python
print(s['Karnataka'])
print(s.xs(2020, level=1))
print(s['Karnataka',2020])
```

```python
df = s.unstack()
print(df)
print(df.stack())
```

```python
import pandas as pd
import numpy as np

r = pd.MultiIndex.from_product([['Maths','Science'],['A','B']])
c = pd.MultiIndex.from_product([['T1','T2'],['Internal','External']])

df = pd.DataFrame(np.random.randint(30,100,(4,4)), index=r, columns=c)
print(df)
```

```python
print(df.loc['Maths'])
print(df.T.groupby(level=0).mean().T)
```

```python
s = pd.Series([1,2,3], index=pd.MultiIndex.from_tuples([('StateB',2020),('StateA',2010),('StateC',2020)]))
try:
    print(s.loc['StateA':'StateB'])
except Exception as e:
    print(e)
print(s.sort_index().loc['StateA':'StateB'])
```

```python
df = pd.DataFrame({'state':['KA','MH'],'year':[2020,2020],'population':[68,125]})
print(df.set_index(['state','year']).reset_index())
```

```python
df = pd.DataFrame({'department':['CSE','AIML','ECE']*4,'gender':['M','F']*6,'salary':range(30000,42000,1000)})
print(pd.pivot_table(df,values='salary',index='gender',columns='department'))
```

```python
print(pd.pivot_table(df,values='salary',index='gender',columns='department',aggfunc='count'))
```

```python
df['age'] = [22,25,35,28,40,45,23,30,38,26,42,48]
df['category'] = pd.cut(df['age'],[0,25,35,100],labels=['Junior','Mid','Senior'])
print(pd.pivot_table(df,values='salary',index='category',columns='department'))
```

```python
print(pd.pivot_table(df,index='department',values=['salary','age'],aggfunc={'salary':'sum','age':'mean'}))
```
